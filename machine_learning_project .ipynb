```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


df=pd.read_csv('hose_price.csv')

def data_visuals(data):
    plt.scatter(data['area'],data['price'])
    plt.xlabel('area')
    plt.ylabel('price')
    plt.title('area vs price')
    plt.show()

def data_evaluation(data):
    print('top 10 points')
    print('-'*40)
    print('-'*40)
    print(data.head(10))
    print('-'*40)
    print('bottom 10 points')
    print(data.tail(10))
    print('-'*40)
    print('-'*40)
    print('shape of data')
    print(data.shape)
    print('-'*40)
    print('data types')
    print(data.dtypes)
    print('-'*40)
    print('Total missing Record :',data.isnull().sum())
    print('-'*40)
    print('summery of data')
    print(data.describe())
    print('Total Duplicate Record :',data.duplicated().sum())
```

```python
def linear_reg(data):
    x=data.drop('price',axis=1)
    y=data['price']
    print('traing  is done on 80% of data')
    x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
    model=LinearRegression()
    model.fit(x_train,y_train)
    predict=model.predict(x_test)
    return x_test,y_test,predict
```

```python
def model_evalution(data):
    x_test,y_test,predict=linear_reg(data)
    print('mean_absolute_error',mean_absolute_error(y_test,predict))
    print('mean_squared_error',mean_squared_error(y_test,predict))
    print('r2_score',r2_score(y_test,predict))

```

```python
def actual_vs_predicted(data):
    x_test,y_test,predict=linear_reg(data)
    plt.scatter(y_test,predict)
    plt.xlabel('actual price')
    plt.ylabel('predicted price')
```

```python
data_evaluation(df)
data_visuals(df)
linear_reg(df)
model_evalution(df)
actual_vs_predicted(df)
```
